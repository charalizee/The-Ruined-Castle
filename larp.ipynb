```python
def print_colored(text, color="reset", style=None):
    # Dictionary mapping color names to their ANSI color codes
    colors = {
        # Normal (Standard) 8 Colors
        "black": "30",
        "red": "31",
        "green": "32",
        "yellow": "33",
        "blue": "34",
        "magenta": "35",
        "cyan": "36",
        "white": "37",
        
        # High-Intensity (Bright) 8 Colors
        "bright_black": "90",
        "bright_red": "91",
        "bright_green": "92",
        "bright_yellow": "93",
        "bright_blue": "94",
        "bright_magenta": "95",
        "bright_cyan": "96",
        "bright_white": "97",
        
        # Reset marker
        "reset": "0"
    }
    
    # Dictionary mapping text styles to their ANSI style codes
    styles = {
        "bold": "1",
        "dim": "2",
        "underline": "4"
    }
    
    # Look up codes (default to empty if not found)
    color_code = colors.get(color.lower(), "0")
    style_code = styles.get(style.lower(), "") if style else ""
    
    # Combine style and color if both are present
    if style_code and color_code != "0":
        ansi_sequence = f"\033[{style_code};{color_code}m"
    elif style_code:
        ansi_sequence = f"\033[{style_code}m"
    else:
        ansi_sequence = f"\033[{color_code}m"
        
    reset_code = "\033[0m"
    
    # Print with the combined style/color and reset at the end
    print(f"{ansi_sequence}{text}{reset_code}")


    # 1. Just color (default behavior)
print_colored("This is standard red text.", color="red")

# 2. Color with bold style
print_colored("This is bold red text!", color="green", style="bold")

# 3. Color with underline style
print_colored("This is underlined green text.", color="green", style="underline")

# 4. Just a style without specifying a color
print_colored(f"This is just underlined text.", style="underline")

print_colored(f"This is just dim text.", style="dim")
```
